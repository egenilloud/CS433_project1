```python
%load_ext autoreload
%autoreload 2
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

import numpy as np
import preprocessing as prep
import matplotlib.pyplot as plt
import implementations as impl
from helpers import load_csv_data
import kNN as knn
```

```python
#loading dataset
x_train, x_test, y_train, train_ids, test_ids = load_csv_data("../data/dataset")
# Load feature names (without first column 'Id')
names = prep.load_feature_names("../data/dataset/x_train.csv")
```

```python
# 1. Split
x_tr, x_va, y_tr, y_va = prep.split_data(x_train, y_train, ratio=0.8, seed=1)

# 2. Preprocessing (fit sur x_tr seulement)
X_tr, X_va, X_te, feature_names, params = prep.preprocess_split(x_tr, x_va, x_test, names)
tx_tr, tx_va, tx_te = prep.add_bias(X_tr, X_va, X_te)
```

## K- Nearest Neigbor Model 

Prediction using Euclidian distance

```python
# Train KNN and evaluate on validation data
y_pred = knn.knn_predict_eucl(X_tr, y_tr, X_va, k=5)

accuracy = np.mean(y_pred == y_va)
print("Validation accuracy:", accuracy)
```

Cross validation to find best hyperparameter k - il faudrait print le score de chaque k 

```python
k_values = np.arange(1, 6, 2)  # 1, 3, 5, ..., 29

scores = knn.knn_cross_validation_f1(
    X_tr, y_tr, k_values, n_folds=5
)

best_index = np.argmax(scores)
best_k = k_values[best_index]

print("Best k:", best_k)
print("Mean validation F1-score:", scores[best_index])
```
